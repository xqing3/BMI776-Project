```python
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt
import seaborn as sns
import os
import liana as li
from liana.method import cellphonedb, cellchat
from tqdm import tqdm
#liana dotplot returns a ggsave object...
from plotnine import ggsave, ggplot

cpdb = sc.read_h5ad("IndividualMiceData/SPF_HF_118_CellPhoneDB.h5ad")
cc = sc.read_h5ad("IndividualMiceData/SPF_HF_118_CellChat.h5ad")
```

```python
x = cc.uns["cpdb_res"]
both = x[x.ligand == ('H2-Aa' or 'H2-Ab1')]
s = sorted(both.source.unique())
li.pl.dotplot(adata = cc,
              colour='lr_probs',
              size='cellchat_pvals',
              inverse_size=True,
              source_labels=[i for i in s if "DC" in i],
              target_labels=['T cells'],
              figure_size=(25, 7),
              ligand_complex=['H2-Aa', 'H2-Ab1'],
              uns_key='cpdb_res'
             )
```
